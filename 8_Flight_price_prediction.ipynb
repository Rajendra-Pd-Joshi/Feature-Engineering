## EDA and Feature Engineering on Flight Price Prediction Dataset

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
df = pd.read_excel('flight_price.xlsx')
df.head()
```

```python
## Get the Basic Info about the data
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
## Feature Engineering Process
df['Day'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
```

```python
df.info()
```

```python
## still day month year are object type so convert them into int
df['Day']=df['Day'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
df.drop('Date_of_Journey',axis = 1, inplace = True)
```

```python
df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df['Arrival_hour']= df['Arrival_Time'].str.split(':').str[0]
df['Arrival_min']= df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head()
```

```python
df['Arrival_hour']= df['Arrival_hour'].astype('int')
df['Arrival_min']= df['Arrival_min'].astype('int')
```

```python
df.drop('Arrival_Time',axis = 1, inplace = True)
```

```python
df.head(2)
```

```python
df['departure_hour'] = df['Dep_Time'].str.split(':').str[0]
df['departure_min'] = df['Dep_Time'].str.split(':').str[1]
```

```python
df.info()
```

```python
df['departure_hour'] = df['departure_hour'].astype(int)
df['departure_min'] = df['departure_min'].astype(int)
```

```python
df.info()
```

```python
df.drop('Dep_Time',axis=1, inplace = True)
```

```python
df.head(2)
```

```python
df['Total_Stops'].unique()
```

```python
df[df['Total_Stops'].isnull()]
```

```python
df['Total_Stops'] = df['Total_Stops'].map({'non-stop':0, '1 stop':1, '2 stops':2, '3 stops':3, '4 stops':4,np.nan:1})
```

```python
df[df['Total_Stops'].isnull()]
```

```python
df.head()
```

```python
df.drop('Route',axis=1, inplace = True)
```

```python
df.head(2)
```

```python
df['Duration_hours'] = df['Duration'].str.extract(r'(\d+)h', expand=False).fillna('0').astype(int)
```

```python
df['Duration_minutes'] = df['Duration'].str.extract(r'(\d+)m', expand=False).fillna('0').astype(int)
```

```python
df.head()
```

```python
df.info()
```

```python
df['Duration_hours'] = df['Duration_hours'].astype(int)
df['Duration_minutes'] = df['Duration_minutes'].astype(int)
df.drop('Duration',axis=1, inplace = True)
```
