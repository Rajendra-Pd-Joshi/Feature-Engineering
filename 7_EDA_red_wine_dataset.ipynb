## Exploratory Data Analysis (EDA) for the dataset red wine

```python
import pandas as pd
df = pd.read_csv(
    'winequality-red.csv'
)

df.head()
```

```python
## Summary of the dataset
df.info()
```

```python
## Generate the descriptive statistics of the dataset
df.describe()
```

```python
df.shape
```

```python
## List down all the columns in the dataset
df.columns
```

```python
df['quality'].value_counts()
```

```python
## Missing values in the dataset

df.isnull().sum()
```

```python
## Checking the Duplicate records 
df[df.duplicated()]
```

```python
## Remove the duplicates 
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
## Correlation
df.corr()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
```

```python
import seaborn as sns
sns.heatmap(df.corr(),annot = True)
```

```python
## Visualization
## Conclusion it is an imbalanced dataset
df['quality'].value_counts().plot(kind ='bar')
plt.xlabel('wine quality')
plt.ylabel('Count')
plt.show()
```

```python
df.head()
```

```python
for column in df.columns:
    sns.histplot(df[column],kde = True)
```

```python
## Univarient and bivarient , multivarient analysis
sns.pairplot(df)
```

```python
## Categorical plot
sns.catplot(x='quality',y='alcohol',data = df, kind = 'box')
```

```python
sns.scatterplot(x='alcohol',y='pH',hue='quality',data = df)
```
